```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
```

```python
path = "lab1_DS.csv"
df = pd.read_csv(path)
name = ['X', 'Y']
for col in df.columns:
  print(f'{col}')
  print(f"Количество: {df[col].count()}")
  print(f'Минимум: {df[col].min()}')
  print(f'Максимум: {df[col].max()}')
  print(f'Среднее: {df[col].mean()}')
```

```python
cols = list(df.columns)
if len(cols) == 2:
  x_name, y_name = cols[0], cols[1]
else:
  while True:
    x = int(input())
    y = int(input())
    if x != y and 0 <= x < len(cols) and 0 <= y < len(cols):
      x_name, y_name = cols[x], cols[y]
      break
```

```python
x = df[x_name].to_numpy()
y = df[y_name].to_numpy()
plt.scatter(x,y)
plt.xlabel(x_name)
plt.ylabel(y_name)
plt.title("Исходные данные")
plt.grid(True)
plt.show()
```

```python
x_mean = x.mean()
y_mean = y.mean()
a = (np.sum((x - x_mean) * (y - y_mean))) / (np.sum((x - x_mean) ** 2))
b = y_mean - a * x_mean
print(f'y = {a:.2f} x + {b:.2f}')
```

```python
x = df[x_name].to_numpy()
y = df[y_name].to_numpy()
plt.scatter(x,y)
x_line = np.linspace(x.min(), x.max(), 100)
y_line = a * x_line + b
plt.plot(x_line, y_line, color = "red")
plt.xlabel(x_name)
plt.ylabel(y_name)
plt.title("Исходные данные")
plt.grid(True)
plt.show()
```

```python
x = df[x_name].to_numpy()
y = df[y_name].to_numpy()
plt.scatter(x,y)
x_line = np.linspace(x.min(), x.max(), 100)
y_line = a * x_line + b
plt.plot(x_line, y_line, color = "red")
y_pred = a * x + b
for xi, yi, yp in zip(x, y, y_pred):
  d = abs(yi - yp)
  if d == 0:
    continue
  y0 = min(yi, yp)
  rect = Rectangle((xi, y0), d, d, facecolor='pink', edgecolor='pink', alpha=0.3, hatch='//')
  plt.gca().add_patch(rect)

plt.xlabel(x_name)
plt.ylabel(y_name)
plt.title("Исходные данные")
plt.grid(True)
plt.xlim(x.min() - 1, x.max() + 1)
plt.ylim(y.min() - 5000, y.max() + 5000)
plt.show()
```
